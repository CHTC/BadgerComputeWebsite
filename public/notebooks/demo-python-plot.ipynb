## 📚 Import data with `pandas`

The data we'll be using is saved in `.csv` files in the `data` directory. We will use the `pandas` package to import the data as DataFrame objects, which are useful to work with.

```python
# Import the pandas package
import pandas as pd
```

```python
# Import Oceania data into a DataFrame
data = pd.read_csv('data/gapminder_gdp_oceania.csv', index_col='country')
print(data)
```

### Modify DataFrames

DataFrames are flexible and easy to manipulate. In the below example, we replace the column title "gdpPercap_\<year\>" with the actual value of `<year>`, in preparation for plotting.

```python
# Replace the column titles with their year
years = data.columns.str.replace('gdpPercap_', '')
data.columns = years.astype(int)
print(data)
```

## 📈 Plot data with `matplotlib`

We will use `matplotlib`, a popular plotting package to visualize the gross domestic product (GDP) of countries in Oceania.


```python
# Import the matplotlib package
from matplotlib import pyplot as plt
```

### Plot the GDP of Australia over time

You can easily plot a subset of your data...

```python
data.loc['Australia'].plot()
```

### Plot the GDP of all countries in Oceania over time

... or you can plot your complete data and even customize it.

You can also save your plot by using the `savefig` function! After running this cell, click the "refresh" icon in the navigation bar to see your plot!

```python
# Plot data and modify labels
data.T.plot()
plt.ylabel('GDP per capita')
plt.xlabel('Year')

# Save figure as a png
fig = plt.gcf()
plt.savefig('gdp_oceania.png')
```

### Customize your plot

Below are a few more options that show you some options for customizing and formatting your plot.
> See the full [matplotlib documentation](https://matplotlib.org/stable/index.html) to see how to customize plots and use advanced features!

```python
# Select two countries' worth of data.
gdp_australia = data.loc['Australia']
gdp_nz = data.loc['New Zealand']

# Plot with differently-colored markers.
plt.plot(years, gdp_australia, 'b-', label='Australia')
plt.plot(years, gdp_nz, 'g-', label='New Zealand')

# Create legend.
plt.legend(loc='upper left')
plt.xlabel('Year')
plt.ylabel('GDP per capita ($)')
plt.title('GDP of Oceania')
plt.savefig('gdp_oceania.png')
```
